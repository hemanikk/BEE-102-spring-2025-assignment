## Q2. Building Markov transition matrix
- INPUT GIVEN : A nucleotide sequence
- OUTPUT GIVEN : A Transition matrix in the form of a dictionary

```python
import pandas as pd
from collections import defaultdict

# DNA sequence input
dna_sequence = "ACGTTGCAAGTCAGAATGCATAGACACAGATACGATACAGAT"
valid_bases = ['A', 'C', 'G', 'T']

# Initialize dictionary to count base-to-base transitions
base_transitions = defaultdict(lambda: defaultdict(int))

# Count each base transition in the sequence
for current_base, next_base in zip(dna_sequence, dna_sequence[1:]):
    if current_base in valid_bases and next_base in valid_bases:
        base_transitions[current_base][next_base] += 1
    else:
        print(f"Invalid base found: {current_base} or {next_base}")

# Normalize counts to get transition probabilities
transition_probabilities = {
    base_from: {
        base_to: base_transitions[base_from][base_to] / sum(base_transitions[base_from].values())
        if sum(base_transitions[base_from].values()) > 0 else 0.0
        for base_to in valid_bases
    }
    for base_from in valid_bases
}

# Create and display the transition matrix as a DataFrame
transition_df = pd.DataFrame(transition_probabilities).T
print("Transition Probability Matrix:")
display(transition_df)
```
