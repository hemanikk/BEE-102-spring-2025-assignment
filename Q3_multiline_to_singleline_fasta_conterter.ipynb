# *Q3. Converting a Multi-line sequence fasta file to single-line fasta file
INPUT GIVEN : A FASTA file with multiline sequences. 
OUTPUT GIVEN : A FASTA file with the multiline sequences converted to single line sequences. 

```python
def convert_multiline_fasta_to_singleline(source_path, destination_path):
    """
    Converts a multi-line FASTA file to a single-line FASTA format.
    """
    with open(source_path, 'r') as input_fasta, open(destination_path, 'w') as output_fasta:
        assembled_sequence = ''

        for record in input_fasta:
            line_content = record.strip()

            if line_content.startswith('>'):
                # If there's a sequence already collected, write it before the next header
                if assembled_sequence:
                    output_fasta.write(assembled_sequence + '\n')
                    assembled_sequence = ''
                output_fasta.write(line_content + '\n')
            else:
                assembled_sequence += line_content

        # Write the last sequence if it exists
        if assembled_sequence:
            output_fasta.write(assembled_sequence + '\n')


# Call the function with actual filenames
convert_multiline_fasta_to_singleline('multiline_input.fasta', 'output.fasta')

print("Formatted FASTA file saved as 'output.fasta' with single-line sequences.")
```
